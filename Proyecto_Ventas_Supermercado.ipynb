# Proyecto 1: Análisis de Ventas y Rentabilidad
**Objetivo:** Analizar los datos históricos de un hipermercado para descubrir qué productos son más rentables, qué segmento de cliente nos genera más ingresos y cuál es la salud general del negocio.

```python
import pandas as pd

# Añadimos la palabra mágica: encoding='latin1'
df = pd.read_csv('/content/Sample - Superstore.csv', encoding='latin1')

# Esto nos mostrará las 5 primeras filas de la tabla
df.head()
```

```python
# 1. Ver el tamaño total de nuestro archivo (filas y columnas)
print("Tamaño de nuestra tabla:", df.shape)

# 2. Buscar agujeros o datos en blanco
print("\n¿Cuántos datos vacíos hay en cada columna?")
print(df.isnull().sum())
```

```python
# Agrupamos las ventas por Categoría y sumamos los beneficios (Profit)
beneficios_categoria = df.groupby('Category')['Profit'].sum().reset_index()

# Lo ordenamos de mayor a menor para que el mejor salga arriba
beneficios_categoria = beneficios_categoria.sort_values(by='Profit', ascending=False)

# Mostramos el resultado
print(beneficios_categoria)
```

```python
import plotly.express as px

# Creamos un gráfico de barras de colores
grafico = px.bar(beneficios_categoria,
                 x='Category',
                 y='Profit',
                 title='Beneficio Económico por Categoría de Producto',
                 color='Category')

# Mostramos el gráfico
grafico.show()
```

```python
# 1. Agrupamos las ventas totales por tipo de cliente (Segment)
ventas_segmento = df.groupby('Segment')['Sales'].sum().reset_index()

# 2. Creamos un gráfico circular tipo "Donut"
grafico_donut = px.pie(ventas_segmento,
                       values='Sales',
                       names='Segment',
                       title='¿Qué tipo de cliente nos genera más Ventas?',
                       hole=0.4) # El hole=0.4 es lo que hace el agujero del donut

# 3. Mostramos el gráfico
grafico_donut.show()
```

```python
# 1. Convertimos la columna de texto a formato de "Fecha real"
df['Order Date'] = pd.to_datetime(df['Order Date'])

# 2. Extraemos solo el "Año" de esa fecha y lo guardamos en una nueva columna
df['Year'] = df['Order Date'].dt.year

# 3. Agrupamos las ventas totales sumadas por cada Año
ventas_año = df.groupby('Year')['Sales'].sum().reset_index()

# 4. Creamos un gráfico de líneas para ver si subimos o bajamos
grafico_tendencia = px.line(ventas_año,
                            x='Year',
                            y='Sales',
                            title='Evolución de las Ventas (¿Estamos creciendo?)',
                            markers=True) # Esto pone un puntito bonito en cada año

# 5. Mostramos la magia
grafico_tendencia.show()
```

### Conclusiones para la Directiva:
* **El motor del negocio:** La Tecnología es el sector que nos deja más beneficio neto. Debemos asegurar el stock de estos productos.
* **El problema oculto:** Se venden muchos Muebles (Furniture), pero el margen de beneficio es minúsculo en comparación. Hay que revisar los costes de envío y almacenamiento urgentemente.
* **Nuestro cliente ideal:** Más del 50% de nuestras ventas vienen del cliente de a pie (Consumer), no de las grandes empresas. Las campañas de publicidad deben centrarse en ellos.
* **Salud de la empresa:** Excelente. Las ventas muestran un crecimiento sólido y continuo durante los últimos 4 años.
